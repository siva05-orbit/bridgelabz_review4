```python
cols= df2.columns
cols_missing = [
    "International_Faculty_Rank",
    "International_Students_Rank",
    "Sustainability_Rank",
    "International_Research_Network_Rank",
    "International_Research_Network_Score",
    "Overall_Score",
    "Sustainability_Score",
    "International_Students_Score",
    "International_Faculty_Score"

]
rank_cols=[
    "International_Faculty_Rank",
    "International_Students_Rank",
    "Sustainability_Rank",
    "International_Research_Network_Rank",
    "Academic_Reputation_Rank",
    "Employment_Outcomes_Rank",
    "Sustainability_Rank",
    "Employer_Reputation_Rank",
    "Faculty_Student_Rank",
    "Citations_per_Faculty_Rank"
]
```

```python
for col in rank_cols:
    df2[col] = df2[col].str.replace("+", "")
    df2[col] = df2[col].str.replace("=", "")
```

```python
for col in cols_missing:
    df2[col] = pd.to_numeric(df2[col], errors="coerce")
    df2[col] = df2[col].fillna(round(df2[col].mean(),1))
```

```python
def clean_range(value):
    
    value = str(value)
    value = value.replace("=", "").replace("+", "")
    if "-" in value:
        
        start= value.split("-")[0]
        end=value.split("-")[1]

        
        return (float(start) + float(end)) / 2
    
    return pd.to_numeric(value, errors="coerce")


rank_cols = [
    "RANK_2024",
    "RANK_2025"
]

for col in rank_cols:
    df2[col] = df2[col].apply(clean_range)
```
